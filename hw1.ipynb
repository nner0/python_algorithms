```python
def sum_all_nums(n):
    total = 0
    for i in range(1, n + 1):
        total += i
    return total
```

```python
sum_all_nums(5)
```

```python
def two_sum(ar, target):
    left = 0
    right = len(ar) - 1
    while left < right:
        cur_sum = ar[right ] + ar[left]

        if cur_sum == target:
            return [left, right]
        elif cur_sum < target:
            left += 1
        else: 
            right -= 1
    return None
```

```python
two_sum([1, 2, 4, 7, 11], 9) 
```

```python
def reverse_ar(ar):
    left = 0
    right = len(ar) - 1

    while left < right:
        ar[left], ar[right] = ar[right], ar[left]
        right -= 1
        left += 1
```

```python
reverse_ar([1, 2, 3, 4, 5])
```

```python
def reverse_part(ar, left, right):
    while left < right:
        ar[left], ar[right] = ar[right], ar[left]
        left += 1
        right -= 1

def rotate(ar, k):
    n = len(ar)
    if n == 0:
        return
    k = k % n

    reverse_part(ar, 0, n - 1)
    reverse_part(ar, 0, k - 1)
    reverse_part(ar, k, n - 1)
```

```python
rotate([1, 2, 3, 4, 5], 2)
```

```python
def merge(ar1, ar2):
    res = []
    i, j = 0, 0
    while i < len(ar1) and j < len(ar2):
        if ar1[i] < ar2[j]:
            res.append(ar1[i])
            i += 1
        else:
            res.append(ar2[j])
            j += 1
    while i < len(ar1):
        res.append(ar1[i])
        i += 1
    while j < len(ar2):
        res.append(ar2[j])
        j += 1
        
    return res
```

```python
merge([1, 3, 5], [2, 4, 6]) 
```

```python
def new_merge(ar1, ar2):
    p1 = len(ar1) - len(ar2) - 1
    p2 = len(ar2) - 1
    p3 = len(ar1) - 1

    while p2 >= 0:
        if p1 >= 0 and ar1[p1] > ar2[p2]:
            ar1[p3] = ar1[p1]
            p1 -= 1
        else:
            ar1[p3] = ar2[p2]
            p2 -= 1
        p3 -= 1
    return ar1
```

```python
new_merge([1, 3, 5, 0, 0, 0], [2, 4, 6])
```

```python
def min_ar(nums, target):
    left = 0
    cur = 0
    min_len = len(nums) + 1

    for right in range(len(nums)):
        cur += nums[right]

        while cur >= target:
            window_size = right - left + 1

            if window_size < min_len:
                min_len = window_size

            cur -= nums[left]
            left += 1
    if min_len == len(nums) + 1:
        return 0

    return min_len
```

```python
min_ar([2, 3, 1, 2, 4, 3], 7)
```

```python
def sort(ar):
    left = 0
    right = len(ar) - 1

    while left <= right:
        if ar[left] == 1:
            ar[left], ar[right] = ar[right], ar[left]
            right -= 1
        else:
            left += 1
    return ar
```

```python
sort([1, 0, 1, 0, 0, 1])
```

```python
def nf(ar):
    low, mid = 0, 0
    high = len(ar) - 1

    while mid <= high:
        if ar[mid] == 0:
            ar[low], ar[mid] = ar[mid], ar[low]
            low += 1
            mid += 1

        elif ar[mid] == 1:
            mid += 1

        elif ar[mid] == 2:
            ar[mid], ar[high] = ar[high], ar[mid]
            high -= 1

    return ar
```

```python
nf([2, 0, 2, 1, 1, 0])
```

```python
def even_first(ar):
    ev_idx = 0
    for i in range(len(ar)):
        if ar[i] % 2 == 0:
            ar[i], ar[ev_idx] = ar[ev_idx], ar[i]
            ev_idx += 1
    return ar
```

```python
even_first([3, 2, 4, 1, 6])
```

```python
def zeros_to_end(ar):
    left = 0

    for i in range(len(ar)):
        if ar[i] != 0:
            ar[i], ar[left] = ar[left], ar[i]
            left += 1
    return ar
```

```python
zeros_to_end([0, 1, 0, 3, 12])
```
