1. Найти корень числа

```python
def my_sqrt(n):
    if n < 0:
        raise ValueError('sqrt must be > 0')
    if n < 2:
        return n
    left = 1
    right = n // 2
    res = 0

    while left <= right:
        mid = (left + right) // 2
        square = mid * mid 

        if square == n:
            return mid
        elif square < n:
            res = mid
            left = mid + 1
        else:
            right = mid - 1
    return res
```

```python
my_sqrt(6)
```

Алгоритм работает за o(log(n)), потому что на каждом шаге работы мы уменьшаем размерность сложности вдвое путем разбиения исследуемой части пополам. Границы мы определяем именно так, поскольку знаем, что для верхней границы справедливо: sqrt(n) < n/2, а для нижней нет смысла рассматривать числа, меньше 1

2. Два ксерокса

```python
def printers(x, y, n):
    if n == 1:
        return min(x, y)
    time = 0
    time += min(x, y)

    left = 0
    right = min(x, y) * (n - 1)

    remaining_copies = n - 1
    best_time = right

    while left <= right:
        mid = (right + left) // 2

        made =  mid // x  + mid // y
        if made >= remaining_copies:
            best_time = mid
            right = mid - 1
        else:
            left = mid + 1
    return time + best_time
```

```python
printers(3,3,4)
```

Достаточность времени проверяем так: T/x + T/y >= n-1

3. Накормить животных

```python
def feed(animals, food):
    animals.sort()
    food.sort()

    animal_p, food_p, fed_cnt = 0, 0, 0
    while animal_p < len(animals) and food_p < len(food):
        if animals[animal_p] <= food[food_p]:
            fed_cnt += 1
            animal_p += 1
            food_p += 1
        else:
            food_p += 1
    return fed_cnt
```

```python
animals = [3, 5, 2, 8]
food = [1, 4, 6]
feed(animals, food)
```

Мы руководствуемся тем, что минимизируем остаток (animal - food). То есть, меньше корма выбрасываем, следовательно, больше достается животным 

4. Найти разницу между двумя строками

```python
def diff(a, b):
    char_counts = {}
    for char in a:
        if char in char_counts:
            char_counts[char] += 1
        else:
            char_counts[char] = 1

    for char in b:
        if char not in char_counts:
            return char
        char_counts[char] -= 1

    
```

```python
a = 'aabbbabababas'
b = 'bbbaaabbaaass'
diff(a, b)
```

Проходимся по строке и делаем хэш-мапу: ключ - символ, значение - сколько раз эта буква встречается в строке. Далее, вторым проходом (уже по b), уменьшаем счетчики символов и в конце выводим единственный не пустой


5. Сумма двух элементов массива

```python
def two_sum(nums, target):
    seen = {}
    for i, cur in enumerate(nums):
        residual = target - cur
        if residual not in seen:
            seen[cur] = i
        else:
            return [seen[residual], i]
    return []
```

```python
nums = [1, 3, 2, 8, 4]
target = 10

two_sum(nums, target)
```

Вместо того, чтобы проходить по массиву двойным циклом, мы смотрим на то, есть ли среди уже просмотренных нами чисел разность текущего числа и нашего таргета

6. Массив анаграмм

```python
def anagramm_array(ar):
    dct = {}
    for word in ar:
        word_sorted = "".join(sorted(word))
        dct.setdefault(word_sorted, [])
        dct[word_sorted].append(word)
    return list(dct.values())
        
```

```python
ar = ["eat","tea","tan","ate","nat","bat"]
anagramm_array(ar)
```
