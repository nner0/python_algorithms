```python
class Node:
    def __init__(self, val):
        self.val = val
        self.next = None
```

```python
def has_cycle(head):
    slow = head
    fast = head

    while fast is not None and fast.next is not None:
        slow = slow.next
        fast = fast.next.next
        if slow is fast:
            return True
    return False
```

```python
def reverse_list(head):
    prev = None
    current = head

    while current is not None:
        next_node = current.next  
        current.next = prev       
        prev = current
        current = next_node

    return prev
```

```python
def middle_node(head):
    slow = head
    fast = head

    while fast is not None and fast.next is not None:
        slow = slow.next
        fast = fast.next.next

    return slow
```

```python
def remove_elements(head, val):
    dummy = Node(0, head)
    prev = dummy
    current = head

    while current is not None:
        if current.val == val:
            prev.next = current.next
        else:
            prev = current
        current = current.next

    return dummy.next
```

```python
def remove_nth_from_end(head, n):
    dummy = Node(0, head)
    slow = dummy
    fast = dummy

    for _ in range(n):
        fast = fast.next
        if fast is None:
            raise ValueError('n больше длины')

    while fast.next is not None:
        slow = slow.next
        fast = fast.next

    slow.next = slow.next.next
    return dummy.next
```

```python
from collections import deque


def is_subsequence_queue(a, b):
    queue = deque(a)

    for letter in b:
        if queue and queue[0] == letter:
            queue.popleft()

    return len(queue) == 0
```

```python
def is_subsequence(a, b):
    i = 0
    j = 0

    while i < len(a) and j < len(b):
        if a[i] == b[j]:
            i += 1
        j += 1

    return i == len(a)
```

```python
def is_palindrome_stack(s):
    stack = []
    for letter in s:
        stack.append(letter)

    for letter in s:
        if letter != stack.pop():
            return False

    return True
```

```python
def is_palindrome_deque(s):
    letters = deque(s)

    while len(letters) > 1:
        left = letters.popleft()
        right = letters.pop()
        if left != right:
            return False

    return True
```

```python
def is_palindrome(s):
    left = 0
    right = len(s) - 1

    while left < right:
        if s[left] != s[right]:
            return False
        left += 1
        right -= 1

    return True
```

```python
def sort_stack(numbers):
    source = numbers.copy()  
    stack = []

    while source:
        current = source.pop()

        while stack and stack[-1] > current:
            source.append(stack.pop())

        stack.append(current)

    return stack
```

```python
def merge_lists(list1, list2):
    dummy = Node(0)
    tail = dummy

    while list1 is not None and list2 is not None:
        if list1.val <= list2.val:
            tail.next = list1
            list1 = list1.next
        else:
            tail.next = list2
            list2 = list2.next
        tail = tail.next

    
    if list1 is not None:
        tail.next = list1
    else:
        tail.next = list2

    return dummy.next
```
